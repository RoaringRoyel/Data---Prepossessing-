```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

# Input data files are available in the read-only "../input/" directory
# For example, running this (by clicking run or pressing Shift+Enter) will list all files under the input directory

import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

```python
# Loading the dataset
data1 = pd.read_csv('/kaggle/input/cid-ids-2017/Monday-WorkingHours.pcap_ISCX.csv')
data2 = pd.read_csv('/kaggle/input/cid-ids-2017/Tuesday-WorkingHours.pcap_ISCX.csv')
data3 = pd.read_csv('/kaggle/input/cid-ids-2017/Wednesday-workingHours.pcap_ISCX.csv')
data4 = pd.read_csv('/kaggle/input/cid-ids-2017/Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv')
data5 = pd.read_csv('/kaggle/input/cid-ids-2017/Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv')
data6 = pd.read_csv('/kaggle/input/cid-ids-2017/Friday-WorkingHours-Morning.pcap_ISCX.csv')
data7 = pd.read_csv('/kaggle/input/cid-ids-2017/Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv')
data8 = pd.read_csv('/kaggle/input/cid-ids-2017/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv')
```

```python
data_list = [data1, data2, data3, data4, data5, data6, data7, data8]

print('Data dimensions: ')
for i, data in enumerate(data_list, start = 1):
  rows, cols = data.shape
  print(f'Data{i} -> {rows} rows, {cols} columns')

data = pd.concat(data_list)
rows, cols = data.shape

print()
print('New dimension:')
print(f'Number of rows: {rows}')
print(f'Number of columns: {cols}')
print(f'Total cells: {rows * cols}')
```

```python
columns_to_drop = ['Flow ID', 'Source Port', 'Source IP', 'Destination IP', 'Timestamp', 'Protocol']

# Drop the specified columns
# Use errors='ignore' in case some columns don't exist in the dataset
data = data.drop(columns=columns_to_drop, errors='ignore')

print(f"Dropped columns: {columns_to_drop}")
print(f"New number of columns: {data.shape[1]}")
```

```python
# Renaming the columns by removing leading/trailing whitespace
col_names = {col: col.strip() for col in data.columns}
data.rename(columns = col_names, inplace = True)
data.columns
```

```python
unique_labels = data['Label'].unique()
print(type(unique_labels))
for i in unique_labels:
  print(i)
```

```python
label_counts = data['Label'].value_counts().sort_values()
print(label_counts)
```

```python
# import numpy as np
# import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Assuming your DataFrame is named 'data' and you have a 'Label' column

# Get the list of unique classes from the 'Label' column
classes = data['Label'].unique()

# 1) Substitute missing values with the class's average value
print("Substituting missing values with class average...")
for col in data.columns:
    if data[col].isnull().any():  # Check if the column has any missing values
        if data[col].dtype in [np.number]: # Only apply to numeric columns
            for cls in classes:
                # Calculate the mean for the current class and column, excluding NaNs
                mean_val = data[data['Label'] == cls][col].mean()
                # Fill NaNs in the current column for the current class
                data.loc[(data['Label'] == cls) & (data[col].isnull()), col] = mean_val
print("Missing value substitution complete.")
print("-" * 30)


# 2) Replace infinite values with the maximum value of the matching class
print("Replacing infinite values with class maximum...")
for col in data.columns:
    if data[col].dtype in [np.number]: # Only apply to numeric columns
        if np.isinf(data[col]).any(): # Check if the column has any infinite values
            for cls in classes:
                # Get the subset of data for the current class and column
                class_data = data[data['Label'] == cls][col]
                # Calculate the max value for the current class and column, excluding infinities and NaNs
                max_val = class_data[np.isfinite(class_data)].max()
                # Replace infinite values (positive and negative) with the calculated max
                data.loc[(data['Label'] == cls) & (np.isinf(data[col])), col] = max_val
print("Infinite value replacement complete.")
print("-" * 30)


# 3) Replace negative values in each feature with the minimum value of its corresponding class
print("Replacing negative values with class minimum...")
for col in data.columns:
     if data[col].dtype in [np.number]: # Only apply to numeric columns
        if (data[col] < 0).any(): # Check if the column has any negative values
            for cls in classes:
                # Get the subset of data for the current class and column
                class_data = data[data['Label'] == cls][col]
                # Calculate the min value for the current class and column, excluding NaNs and infinities
                min_val = class_data[np.isfinite(class_data)].min()
                # Replace negative values with the calculated min for the specific class
                data.loc[(data['Label'] == cls) & (data[col] < 0), col] = min_val
print("Negative value replacement complete.")
print("-" * 30)

# 4) Normalize the dataset features using unity-based normalization
print("Normalizing features using Unity-Based Normalization...")
# Exclude the 'Label' column from normalization
features_to_normalize = data.columns.tolist()
if 'Label' in features_to_normalize:
    features_to_normalize.remove('Label')

scaler = MinMaxScaler()

# Fit the scaler on the selected features and transform the data
data[features_to_normalize] = scaler.fit_transform(data[features_to_normalize])
print("Normalization complete.")
print("-" * 30)

print("Preprocessing steps completed.")
# You can now inspect the data to see the changes, e.g., data.head(), data.describe()
```
